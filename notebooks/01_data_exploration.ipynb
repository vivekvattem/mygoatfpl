# FPL AI Predictor — Phase 1 Exploration

## 1. Imports

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
PROCESSED = PROJECT_ROOT / 'data' / 'processed'
MIN_MINUTES = 60
```

## 2. Load processed data

```python
players = pd.read_csv(PROCESSED / 'players.csv')
fixtures = pd.read_csv(PROCESSED / 'fixtures.csv')
fixture_summary_3gw = pd.read_csv(PROCESSED / 'fixture_summary_3gw.csv')
fixture_summary_5gw = pd.read_csv(PROCESSED / 'fixture_summary_5gw.csv')
```

## 3. Dataset overview

```python
display(players.head())
players.info()
display(players.describe(include='all').T)
```

## 4. Top players by total points

```python
display(players.nlargest(15, 'total_points')[['player', 'team_name', 'position', 'price', 'total_points']])
```

## 5. Top midfielders by xGI/90

```python
eligible = players.query('minutes >= @MIN_MINUTES')
display(eligible.query("position == 'MID'").nlargest(15, 'xGI_per90')[['player', 'team_name', 'minutes', 'xGI_per90']])
```

## 6. Top forwards by xGI/90

```python
display(eligible.query("position == 'FWD'").nlargest(15, 'xGI_per90')[['player', 'team_name', 'minutes', 'xGI_per90']])
```

## 7. Best value attackers

```python
attackers = eligible.query("position in ['MID', 'FWD']")
display(attackers.nlargest(15, 'attacking_value_score')[['player', 'position', 'price', 'attacking_score', 'attacking_value_score']])
```
