## 6. Position-level rankings

```python
for position in ['GK', 'DEF', 'MID', 'FWD']:
    display(predictions.query('position == @position').nlargest(10, 'display_xpts'))
```

## 7. Availability adjustments

```python
display(predictions.query('display_xpts != availability_adjusted_xpts')[['player', 'status', 'chance_of_playing_next_round', 'display_xpts', 'availability_adjusted_xpts']])
```

## 8. Uncertainty

```python
display(predictions.nlargest(20, 'display_xpts')[['player', 'display_xpts', 'xpts_lower', 'xpts_upper']])
```

## 10. Current limitations

Current predictions are compressed toward average outcomes and are not reliable estimates of explosive 10+ point scores. Availability adjustment uses only official published percentages, uncertainty bands are empirical, and no transfer, lineup, bench, or captain optimization is performed.
